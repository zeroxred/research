# Research Plan

**Title: Robust Temporal Representation Learning for Weakly Supervised Video Anomaly Detection**

Video Anomaly Detection (VAD) has become an important research area due to its applications in intelligent surveillance, public safety, industrial inspection, and autonomous systems. Despite remarkable progress in recent years, existing methods still struggle with learning robust temporal representations, generalizing to unseen anomalies, and efficiently modeling long video sequences. These challenges become even more pronounced in Weakly Supervised Video Anomaly Detection (WSVAD), where only video-level labels are available during training.

The objective of this research is to investigate how robust temporal representations can improve anomaly detection in long and complex videos. Rather than focusing on a particular architecture, this work addresses the broader problem of temporal representation learning and explores how recent advances in State Space Models (SSMs), Vision-Language Models (VLMs), and efficient sequence modeling can contribute to more accurate, robust, and computationally efficient anomaly detection systems.

Despite the success of Transformer-based and Vision-Language approaches, there remains limited understanding of how temporal representations should be learned to simultaneously achieve long-range dependency modeling, computational efficiency, and robust generalization to unseen anomalies. The central hypothesis of this research is that efficient temporal representations, enriched by semantic information from Vision-Language Models, can significantly improve the robustness and generalization capability of WSVAD systems.

The first stage of the research will consist of a comprehensive study of existing VAD methods, including CNN-based, Transformer-based, and Vision-Language approaches. Particular attention will be given to understanding the strengths and limitations of temporal modeling strategies employed by recent methods such as RTFM, VadCLIP, OV-VAD, and related work.

The second stage will investigate State Space Models, particularly the Mamba architecture and its variants, as alternatives to Transformer-based temporal encoders. Since the original Mamba architecture is inherently causal and unidirectional, this research will investigate whether bidirectional State Space Models or other temporal modeling strategies provide more effective representations for offline video anomaly detection. Rather than simply replacing existing temporal modules, the research will analyze the temporal properties captured by SSM-based approaches and identify the conditions under which they outperform attention-based models.

The final stage will integrate semantic information from Vision-Language Models with efficient temporal representations to improve the recognition of unseen or weakly defined anomalies. Experimental evaluation will be conducted on benchmark datasets such as UCF-Crime, XD-Violence, and ShanghaiTech using frame-level AUC, average precision, computational efficiency, and robustness as evaluation metrics.

Ultimately, this research aims to advance the understanding of temporal representation learning for Weakly Supervised Video Anomaly Detection. The expected outcome is a principled approach to combining efficient sequence modeling with semantic information, contributing both theoretical insights and practical methods for developing more robust, scalable, and generalizable video anomaly detection systems.
