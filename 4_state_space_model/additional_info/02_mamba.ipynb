### 3.2 Improving SSMs with Selection

Consider an input sequence

$$
x \in \mathbb{R}^{B \times L \times D},
$$

where:

- $B$: batch size.
- $L$: sequence length.
- $D$: feature (embedding) dimension.
- $N$: hidden state dimension of the SSM.

The main idea of Mamba is to make the SSM parameters depend on the input, allowing the model to dynamically decide how information should be processed.

In **S4**, the parameters are fixed throughout the entire sequence:

$$
A,\; B,\; C,\; \Delta = \text{learnable parameters}
$$

where:

- $A$: state transition matrix, controlling how the hidden state evolves.
- $B$: input projection matrix, determining how the input updates the hidden state.
- $C$: output projection matrix, mapping the hidden state to the output.
- $\Delta$: discretization step size.

After discretization,

$$
(\overline{A}, \overline{B}) = \mathrm{discretize}(A, B, \Delta),
$$

the same dynamics are applied at every time step. Therefore, S4 is **time-invariant** and can be efficiently implemented as either a recurrence or a convolution.

In **Mamba (Selective SSM)**, the parameters become functions of the current input:

$$
B = s_B(x), \qquad
C = s_C(x), \qquad
\Delta = s_\Delta(x),
$$

where

$$
s_B(x)=\mathrm{Linear}_N(x), \qquad
s_C(x)=\mathrm{Linear}_N(x),
$$

and

$$
s_\Delta(x)=\mathrm{Broadcast}_D(\mathrm{Linear}_1(x)).
$$

The resulting value of $\Delta$ is passed through a Softplus activation,

$$
\tau_\Delta = \mathrm{softplus},
$$

ensuring that $\Delta > 0$.

Since these parameters are generated for every input token, their shapes change from

$$
B,C:(D,N)
$$

to

$$
B,C:(B,L,N),
$$

and

$$
\Delta:(D)
\rightarrow
(B,L,D).
$$

For example, if a sequence contains $L=100$ tokens, the model generates 100 different sets of parameters:

$$
\{B_1,\ldots,B_{100}\}, \qquad
\{C_1,\ldots,C_{100}\}, \qquad
\{\Delta_1,\ldots,\Delta_{100}\}.
$$

Thus, each token is processed with its own state-space dynamics instead of sharing the same parameters across the entire sequence.
This makes the model **time-varying**, meaning each position in the sequence has its own state-space dynamics. Consequently, the convolution formulation is no longer valid, and the sequence must be processed through a recurrent scan.

### Memory-Efficient Scan

Instead of storing all input-dependent SSM parameters in GPU **HBM (High Bandwidth Memory)**, Mamba processes the sequence in small chunks using **SRAM (Shared Memory)**, which is much faster but has limited capacity.

For each chunk, the model:

1. Loads the required SSM parameters from **HBM** to **SRAM**.
2. Performs **discretization**.
3. Executes the **recurrent scan** entirely in **SRAM**.
4. Writes only the final outputs back to **HBM**.
5. Repeats the process for the next chunk while carrying the hidden state.

```text
HBM (large, slow)
      │
      ▼
Load a chunk to SRAM
      │
      ▼
SRAM (small, fast)
 ├── Discretization
 ├── Recurrent Scan
 ├── Update Hidden State
 └── Compute Outputs
      │
      ▼
Write outputs to HBM
      │
      ▼
Repeat for the next chunk
```

Since only one chunk is processed at a time, the entire sequence never needs to fit into SRAM. This significantly reduces memory traffic between HBM and SRAM, improving GPU efficiency.

Each projection contains

$$

$$

$$

Substituting the values

$$

$$

$$

gives

$$

=

Since

$$

$$

we obtain

$$
ED^2
=

=
524,288.
$$

This is exactly the number of parameters in **one** linear projection.

Since there are **three** linear projections, the total is

$$
3ED^2
=
3 \times 524,288
=
1,572,864.

### Key Insight

The paper expresses the parameter count as

$$
3ED^2
$$

because it generalizes to **any model size**.

For example, if a larger model uses

- $D = 768$
- $E = 4$

the same formula still applies without repeating the entire calculation.

Thus, the expression

$$
3ED^2
$$

simply represents the total number of parameters contributed by the **three linear projections** of a Mamba block.

### Why Does Mamba Use Two Blocks?

The paper states:

> *"We always fix to \(E = 2\) in our experiments and use two stacks of the block to match the \(12D^2\) parameters of a Transformer's interleaved MHA and MLP blocks."*

The goal is to make the comparison **fair**. Instead of comparing models with different numbers of parameters, the authors keep approximately the same parameter budget.

---

### Parameters in One Mamba Block

From the previous section, one Mamba block contains three linear projections:

- Main input projection
- Gate input projection
- Output projection

Each projection contributes

$$
ED^2
$$

parameters.

Therefore, one Mamba block has

$$
3ED^2.
$$

Since Mamba fixes

$$
E = 2,
$$

we obtain

$$
3ED^2
=
3(2D^2)
=
6D^2.
$$

Thus, **one Mamba block contains approximately**

$$
6D^2
$$

parameters.

### Parameters in a Transformer Block

A standard Transformer layer consists of two sub-blocks:

```text
Multi-Head Attention (MHA)
↓

MLP
```

Ignoring biases and LayerNorm parameters, each contributes approximately

$$
6D^2.
$$

Therefore,

```text
MHA
≈ 6D²

+

MLP
≈ 6D²
```

giving a total of

$$
12D^2.
$$

### Why Stack Two Mamba Blocks?

Since one Mamba block contains

$$
6D^2,
$$

stacking two of them gives

$$
2 \times 6D^2
=
12D^2.
$$

Therefore,

```text
Mamba Block
≈ 6D²

+

Mamba Block
≈ 6D²

----------------

Total
≈ 12D²
```

which matches the parameter count of

```text
Transformer

Multi-Head Attention
≈ 6D²

+

MLP
≈ 6D²

----------------

Total
≈ 12D²
```

## 3.51 Connection to Gating Mechanisms

This section shows that Mamba's selective mechanism is a generalization of the gating mechanism used in classical RNNs (e.g., GRUs/LSTMs). Under a simplified SSM configuration, the selective recurrence reduces exactly to a gated RNN update, providing a theoretical foundation for interpreting the input-dependent discretization parameter Δ as a generalized gate.

### 4.2 Scaling Laws

Scaling laws describe how a model's performance changes as computational resources increase (e.g., model size, training data, or FLOPs). A model is said to **scale well** if its performance continues to improve efficiently as more compute is invested.

> ***Perplexity (PPL)** is the standard evaluation metric for language models. It measures how well a model predicts the next token in a sequence; **lower perplexity indicates better predictive performance** and, therefore, a stronger language model.*

In Figure 4, the x-axis represents the total training compute (FLOPs), while the y-axis shows **perplexity**. Both axes are plotted on a logarithmic scale because the compared models span several orders of magnitude in computational cost.

The results show that **Mamba consistently reduces perplexity as training compute increases**, demonstrating a more favorable scaling behavior than previous attention-free architectures such as Hyena, H3++, RWKV, and RetNet. At a sequence length of **2048**, Mamba achieves performance comparable to the strongest modern Transformer baseline (Transformer++). More importantly, at **8192** tokens, Mamba maintains this advantage, whereas other attention-free models fall further behind.

This makes Mamba the **first attention-free architecture** to match the performance of a highly optimized Transformer while retaining linear-time complexity. The results suggest that Mamba utilizes additional compute more effectively and remains robust as the context length increases.
