# Weakly supervised learning

A major disadvantage of traditional supervised learning techniques is their reliance on large amounts of hand-labeled training data. Weakly supervised learning addresses this limitation by leveraging lower-quality or less precise labels provided at a higher level of abstraction. These labels are generally easier and less expensive to obtain, while still enabling models to achieve performance comparable to that of fully supervised approaches.

For example, a surveillance video can be labeled as anomalous without identifying the specific temporal segments containing the anomaly. Another example is image classification, where an image is labeled according to the objects it contains without providing bounding boxes or pixel-level annotations. These forms of weak supervision substantially reduce annotation costs while retaining useful information for learning.

# Multiple instance learning

Multiple Instance Learning (MIL) is a weakly supervised learning paradigm in which data samples are grouped into collections known as bags. Instead of assigning labels to each individual instance, annotations are provided only at the bag level. Under the standard MIL assumption, a bag is considered negative when all of its instances belong to the negative class, whereas it is considered positive when at least one instance belongs to the positive class.
