## Hinge Loss (SVM)

Hinge Loss is the loss function used by Support Vector Machines (SVMs). It encourages not only correct classification but also a sufficient margin between classes.

$$
L(x,y)=\max(0,\;1-yf(x))
$$

where:
- $y \in \{-1,+1\}$ is the true label.
- $f(x)$ is the model score.

Properties:
- Loss is zero when a sample is correctly classified with a margin of at least 1.
- Correct predictions close to the decision boundary are still penalized.
- Misclassified samples receive a larger penalty.

The hinge loss promotes large-margin separation between classes.

Si:

- $y=+1$ (positivo), queremos f(x) grande.
- $y=−1$ (negativo), queremos f(x) muy negativo.

Por ejemplo:

$$
y=1,f(x)=2
$$

$$
1−yf(x)=1−2=−1
$$

Loss = 0.

### Appendix — Graph-Based Modeling in Video Anomaly Detection (Zhong et al., 2019)

**Paper:** *Graph Convolutional Label Noise Cleaner for Weakly-Supervised Video Anomaly Detection*  
**Authors:** Jiqing Zhong, Nian Liu, Feifei Li, Weiming Liu, Shenghua Gao (CVPR 2019)

This work introduced graph-based modeling into Weakly-Supervised Video Anomaly Detection (WSVAD). Instead of treating video segments independently, the method represents each video segment as a node in a graph and connects related segments based on feature similarity. A Graph Convolutional Network (GCN) is then used to propagate information between connected nodes, allowing semantically similar segments to influence each other even when they are far apart in time.

**Example:**

Consider a surveillance video divided into segments:

```text
Segment 1  → people walking
Segment 2  → people walking
Segment 3  → fighting
Segment 4  → people walking
Segment 5  → fighting
```

Although Segments 3 and 5 are temporally distant, their visual features may be very similar because both contain fighting events. The graph can therefore connect these segments:

```text
Segment 3 (fighting)
        ↕
Segment 5 (fighting)
```

During graph convolution, information is exchanged between connected nodes, reinforcing their anomalous characteristics and helping the model distinguish them from normal segments.

**Key idea:**
- Represent video segments as graph nodes.
- Connect nodes according to feature similarity.
- Use GCNs to propagate information across related segments.
- Improve anomaly detection by leveraging global relationships and reducing the effect of noisy supervision.

**Relevance to VadCLIP:**
VadCLIP inherits the graph-based idea introduced by Zhong et al. for its Global Temporal Adapter. However, instead of using GCNs primarily for label-noise cleaning, VadCLIP uses them to capture long-range temporal and semantic relationships between CLIP-based video features.

### Appendix — Graph Convolutional Networks (GCNs) for Video Understanding

Graph Convolutional Networks (GCNs) can be used in video understanding to model relationships between frames or video segments. Instead of treating a video strictly as a sequence, a GCN represents it as a graph where each frame or segment becomes a node, and edges define relationships between them.

The main idea is that events occurring in different parts of a video may still be related, even if they are far apart in time. By constructing a graph, information can be exchanged between semantically or temporally related frames, allowing the model to capture broader context than local temporal operations alone.

A graph for video analysis consists of:

- **Nodes:** frame-level or segment-level video features.
- **Edges:** relationships between nodes, often based on:
  - Feature similarity
  - Temporal proximity
  - Learned relationships

During graph convolution, each node updates its representation by aggregating information from its connected neighbors:

```text
Updated Frame Representation
=
Own Features
+
Neighbor Features
```

As multiple graph convolution layers are applied, information propagates through increasingly distant parts of the video.

**Example**

Consider a video containing the following segments:

```text
Segment 1 → Person entering a room
Segment 2 → Person walking
Segment 3 → Person sitting
Segment 4 → Person standing up
Segment 5 → Person leaving the room
```

A sequential model would mainly focus on adjacent relationships:

```text
S1 → S2 → S3 → S4 → S5
```

A graph-based model can additionally connect semantically related segments:

```text
S1 (entering)
      ↕
S5 (leaving)

S2 (walking)
      ↕
S4 (standing up)
```

Even though these segments are temporally distant, they may represent related stages of the same activity. Through graph convolution, information can flow between them, helping the model build a more complete understanding of the video.

**Why Use GCNs in Video?**

- Capture long-range dependencies between distant frames.
- Connect semantically similar events regardless of temporal distance.
- Incorporate both temporal and feature-based relationships.
- Provide global context without requiring full self-attention over the entire sequence.

**Comparison with Other Video Modeling Approaches**

| Method | Main Focus |
|----------|----------|
| 3D CNNs | Local spatiotemporal patterns |
| Transformers | Global attention across frames |
| GCNs | Information propagation through graph relationships |

**Key Intuition**

Rather than asking *"which frames are nearby?"*, GCNs ask *"which frames are related?"*. This allows video representations to incorporate contextual information from both temporally close and semantically similar parts of the video.
