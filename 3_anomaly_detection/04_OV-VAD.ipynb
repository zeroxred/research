## Introduction

### What is Video Anomaly Detection (VAD)?

Video Anomaly Detection (VAD) aims to identify unusual events that do not conform to expected patterns in videos. It has become increasingly important due to its applications in intelligent surveillance systems, public safety monitoring, and video content review.

Over the years, VAD research has mainly focused on two paradigms: **Semi-Supervised VAD** and **Weakly-Supervised VAD**.

## Semi-Supervised VAD

### Figure 1(a)

During training, only normal videos are available.

The goal is to learn the normality pattern and, during inference, any behavior that deviates from this learned pattern is considered an anomaly.

From a machine learning perspective, this can be viewed as a **one-class classification** problem.

Common approaches include:

- One-class classifiers.
- Frame reconstruction.
- Frame prediction.
- Jigsaw puzzles and other self-supervised learning tasks.

### Example

If a model is trained only on videos showing normal traffic flow, a car accident will be detected as an anomaly because it does not resemble anything observed during training.

---

## Weakly-Supervised VAD

### Figure 1(b)

During training, both normal and abnormal videos are available.

However, abnormal events are not temporally annotated. The model only knows whether a video contains an anomaly or not.

This problem can be viewed as a **binary classification** task:

- Normal
- Abnormal

Popular approaches include:

- Multiple Instance Learning (MIL).
- Top-K mechanisms.

### Example

A video may be labeled as "abnormal", but the exact timestamp of the abnormal event is unknown. The model must discover by itself which video segments are responsible for the abnormal label.

---

## Limitation of Traditional VAD Methods

Although these approaches have achieved strong performance on several benchmarks, they share an important limitation:

> They can only detect anomaly categories that are known during training.

In other words, they operate under a **closed-set** setting.

### Example

If a model is trained to recognize:

- Fighting
- Shooting

but has never seen:

- Traffic Accident

it may fail to detect the accident correctly.

The authors point out that this limitation is particularly problematic in real-world deployments, where new anomaly types continuously emerge and are often absent from the training data.

---

## Open-Set VAD

### Figure 1(c)

To overcome the limitations of closed-set anomaly detection, recent works have explored **Open-Set VAD**.

The idea is to train a model using:

- Normal videos.
- Some known anomaly categories (*seen anomalies*).

and expect it to detect completely new anomaly categories (*unseen anomalies*) during testing.

### Example

Training:

- Fighting
- Shooting

Testing:

- Traffic Accident

The goal is for the model to still recognize the accident as an anomaly even though no accident examples were available during training.

---

## Limitation of Open-Set VAD

Although Open-Set VAD improves the ability to generalize to unseen anomalies, it still suffers from an important limitation:

> It can detect that something is abnormal, but it cannot clearly understand or describe what type of anomaly it is.

As a result, anomaly detection remains semantically ambiguous.

---

## Motivation for Open-Vocabulary VAD (OV-VAD)

Recent advances in large-scale vision-language pre-trained models have demonstrated strong transfer learning capabilities and rich cross-modal semantic knowledge.

Leveraging these capabilities, the authors introduce a new task:

## Open-Vocabulary Video Anomaly Detection (OV-VAD)

### Figure 1(d)

OV-VAD aims to:

1. Detect anomalies.
2. Categorize anomalies.
3. Handle known (*seen/base*) anomalies.
4. Handle unknown (*unseen/novel*) anomalies.

### Example

Given a video containing a traffic accident, the system should not only output:

> "Anomaly detected"

but ideally:

> "Crash"

even if accident examples were never observed during training.

---

## Challenges of OV-VAD

The authors identify two major challenges.

### 1. Detection and Categorization

The system must not only determine whether an event is normal or abnormal, but also assign a meaningful semantic category to the detected anomaly.

### 2. Generalization to Unseen Anomalies

The model must be capable of handling both:

- Seen anomalies.
- Unseen anomalies.

---

## Proposed Solution

The authors decompose OV-VAD into two complementary sub-tasks.

### A. Class-Agnostic Detection

The goal is to determine whether an event is normal or abnormal without considering its specific category.

To improve this task, the authors introduce:

#### Temporal Adapter (TA)

A nearly weight-free module designed to model temporal relationships between video frames.

#### Semantic Knowledge Injection (SKI)

A module that injects textual knowledge from large language models into visual representations.

The goal is to enrich visual features with semantic information.

---

### B. Class-Specific Categorization

Once an anomaly is detected, the model attempts to identify its category.

Inspired by CLIP, the authors use a scalable approach based on aligning:

- Textual labels.
- Video representations.

#### Novel Anomaly Synthesis (NAS)

The authors additionally generate synthetic visual materials (images and videos) to help the model learn unseen anomaly categories.

The purpose is to improve the classification of novel anomalies.

---

## Evolution of VAD Paradigms

| Paradigm | Detect Seen Anomalies | Detect Unseen Anomalies | Categorize Anomalies |
|-----------|-----------|-----------|-----------|
| Semi-Supervised VAD | ✓ | ✗ | ✗ |
| Weakly-Supervised VAD | ✓ | ✗ | ✗ |
| Open-Set VAD | ✓ | ✓ | ✗ |
| Open-Vocabulary VAD | ✓ | ✓ | ✓ |

---

## Key Takeaway

The central contribution of this work is the transition from traditional closed-set anomaly detection toward **open-vocabulary anomaly understanding**, where a model can not only detect both seen and unseen anomalies but also provide semantic descriptions of what those anomalies are by leveraging knowledge from large vision-language pre-trained models.

# Weakly-Supervised VAD vs Open-Set VAD

A common misconception is that Weakly-Supervised VAD (WSVAD) and Open-Set VAD solve exactly the same problem, since both can sometimes detect anomalies that were not present during training.

## Weakly-Supervised VAD (MIL-based Methods)

In WSVAD, the model is trained using:

* Normal videos.
* Abnormal videos.

However, only video-level labels are available. The temporal locations of anomalies are unknown.

Typical methods such as MIL learn to distinguish between normal and abnormal patterns by exploiting the differences between positive and negative bags.

### Example

Training:

* Fighting
* Shooting

Testing:

* Fighting
* Shooting

This is a **closed-set setting** because the anomaly categories present during testing are the same as those seen during training.

Although MIL may occasionally detect unseen anomalies such as traffic accidents or explosions, this behavior is merely a side effect of learned anomaly representations rather than an explicit design objective.

The model often learns:

> "This looks similar to the abnormal events I saw during training."

rather than

> "This is abnormal even though I have never seen it before."

---

## Open-Set VAD

Open-Set VAD explicitly focuses on detecting unseen anomaly categories.

Training:

* Fighting
* Shooting

Testing:

* Traffic Accident
* Fire
* Explosion

The goal is to detect these novel anomalies despite having no training examples for them.

Instead of relying heavily on specific anomaly categories, Open-Set VAD attempts to learn more general abnormality concepts and reduce dependence on seen classes.

The model aims to answer:

> "Is this abnormal compared to normal behavior?"

rather than

> "Does this resemble the anomalies I have seen before?"

---

## Why Is WSVAD Considered Closed-Set?

Most WSVAD benchmarks assume:

Training classes:

* Fighting
* Shooting
* Robbery

Testing classes:

* Fighting
* Shooting
* Robbery

Since the anomaly categories remain unchanged, the problem is considered closed-set.

Open-Set VAD introduces a more challenging evaluation where test anomalies belong to categories never observed during training.

---

## Intuitive Analogy

Imagine training a security guard.

### WSVAD (MIL)

You show the guard:

* 100 fights.
* 100 shootings.

Later, the guard sees a traffic accident.

The guard may think:

> "This does not look like a fight or a shooting."

and fail to react appropriately.

### Open-Set VAD

Instead of memorizing specific anomaly categories, the guard learns broader concepts of abnormal and dangerous situations.

When seeing a traffic accident, the guard thinks:

> "I have never seen this before, but it is clearly abnormal."

---

## Relation to OVVAD

Open-Vocabulary Video Anomaly Detection (OVVAD) goes one step further.

It aims not only to:

* Detect seen anomalies.
* Detect unseen anomalies.

but also to:

* Categorize them semantically.

### Example

Instead of producing:

> "Anomaly detected"

the model should output:

> "Traffic Accident"

even if no traffic accident examples were available during training.

Therefore, the progression can be summarized as:

| Paradigm              | Detect Seen Anomalies | Detect Unseen Anomalies | Categorize Anomalies |
| --------------------- | --------------------- | ----------------------- | -------------------- |
| Semi-Supervised VAD   | ✓                     | ✗                       | ✗                    |
| Weakly-Supervised VAD | ✓                     | ✗                       | ✗                    |
| Open-Set VAD          | ✓                     | ✓                       | ✗                    |
| Open-Vocabulary VAD   | ✓                     | ✓                       | ✓                    |

## Key Insight

MIL-based methods may sometimes detect unseen anomalies, but this capability is not explicitly optimized during training. Open-Set VAD is specifically designed and evaluated for generalization to novel anomaly categories, while OVVAD additionally seeks to provide semantic descriptions of those anomalies.


# Related Work – Key Observations

## Semi-Supervised VAD

- Learns only normal patterns.
- Main approaches:
  - Reconstruction-based.
  - Prediction-based.
  - One-class learning.
- Detects anomalies as deviations from normality.

---

## Weakly-Supervised VAD

- Uses both normal and abnormal videos.
- Usually formulated as a binary classification problem.
- Major developments:
  - MIL (Sultani et al., 2018).
  - GCN (Zhong et al., 2019).
  - Attention and temporal modeling.
  - Audio-visual VAD.
  - Transformers and pseudo-label learning.

---

## CLIP-based WSVAD

### Mainly used CLIP as a visual feature extractor

Most CLIP-based VAD methods only use the image encoder to obtain stronger visual representations.

Typical pipeline:

```text
Video Frames
      ↓
CLIP Image Encoder
      ↓
Visual Features
      ↓
MIL / Classification Network
      ↓
Anomaly Score
```

In this setting, CLIP is essentially treated as a replacement for traditional feature extractors such as:

- C3D
- I3D
- Video Swin Transformer

The textual branch of CLIP is largely ignored.

---

### Did not fully exploit CLIP's zero-shot or open-vocabulary capabilities

One of CLIP's main strengths is that it learns a shared embedding space between images and text.

This allows CLIP to perform zero-shot recognition by matching visual features with textual descriptions.

Example:

```text
Text:
    "fighting"
    "shooting"
    "traffic accident"

Image/Video
    ↓

Similarity Matching
```

Even without training examples of a specific category, CLIP can often associate visual content with the correct textual concept.

However, previous CLIP-based VAD methods generally do not use CLIP to answer:

```text
What anomaly is this?
```

Instead, they use CLIP only to answer:

```text
Is this anomalous?
```

As a result, they inherit stronger visual features from CLIP but do not leverage:

- Zero-shot recognition.
- Open-vocabulary classification.
- Semantic anomaly understanding.
- Text-guided anomaly categorization.

This is the gap that OVVAD attempts to address.

## Open-Set VAD

- Trains with normal and seen anomalies.
- Aims to detect unseen anomalies.
- Focuses on open-world anomaly detection.
- Does not explicitly categorize anomaly types.

---

## OVVAD Motivation

Previous works focus on:

```text
Is this anomalous?
```

OVVAD focuses on:

```text
Is this anomalous?
What anomaly is it?
```

The key transition is:

```text
Semi-Supervised VAD
    ↓
Weakly-Supervised VAD
    ↓
Open-Set VAD
    ↓
Open-Vocabulary VAD
```

where the goal evolves from learning normality, to detecting unseen anomalies, and finally to semantically categorizing them.

## Problem Statement

The training set is defined as:

$$
X = \{x_i\}_{i=1}^{N+A}
$$

where:

- $X_n$ is the set of normal samples $X_n = \{x_i\}_{i}^{N}$.
- $X_a$ is the set of abnormal samples $X = \{x_i\}_{i=N+1}^{N+A}$.

Each abnormal video has a **video-level category label**:

$$
y_i \in C_{base}
$$

where $C_{base}$ represents the set of **base (seen) anomaly categories**.

The complete category set is defined as:

$$
C = C_{base} \cup C_{novel}
$$

where $C_{novel}$ denotes the set of **novel (unseen) anomaly categories**.

### Objective

Given the training set $X$, the model aims to:

1. Predict an **anomaly confidence score** for every frame.
2. Identify the **anomaly category** if an anomaly is present in the video.

The model should be able to detect and categorize both **base (seen)** and **novel (unseen)** anomalies.

## 3.1 Overall Framework

Traditional closed-set VAD methods are not suitable for open-vocabulary scenarios. Therefore, the authors build their framework on **CLIP**, leveraging its **zero-shot generalization ability**.

Given a training video:

1. The video is fed into CLIP's image encoder to extract **frame-level features**.
2. These features are processed by:
   - **Temporal Adapter (TA)**
   - **Semantic Knowledge Injection (SKI)**
   - **Detector**
3. The detector outputs a **frame-level anomaly confidence score**. This branch is responsible for **class-agnostic detection**.

For **class-specific categorization**:

1. Frame-level features are aggregated into a **video-level feature**.
2. Textual embeddings are generated for anomaly categories.
3. The anomaly category is predicted by aligning the video-level feature with the textual embeddings using a **cross-modal alignment mechanism**.

Additionally, the authors introduce the **Novel Anomaly Synthesis (NAS)** module, which uses **LLMs** and **AIGC models** to generate potential novel anomalies and improve the identification of unseen anomaly categories.

## 3.2 Temporal Adapter (TA) Module

CLIP's image encoder extracts strong visual features but does not model **temporal dependencies**, since it is pre-trained on image-text pairs.

Although temporal transformers are commonly used to capture temporal relationships, they may reduce performance on **novel categories** because their additional parameters can overfit to the training data.

To address this issue, the authors propose a **nearly weight-free Temporal Adapter (TA)** based on **Graph Convolutional Networks (GCNs)**.

The Temporal Adapter is defined as:

$$
x_t = LN(\text{softmax}(H)x_f)
$$

where:

- $x_f$ are the frame-level features.
- $H$ is the adjacency matrix.
- $LN$ is Layer Normalization.

The adjacency matrix is computed as:

$$
H(i,j)=\frac{-|i-j|}{\sigma}
$$

where $\sigma$ is a hyperparameter controlling the influence range of temporal distance.

The proximity between two frames depends only on their relative temporal positions: frames that are closer in time have higher proximity, while frames farther apart have lower proximity.

The only learnable parameters in the TA module are those in the Layer Normalization.

## 3.3 Semantic Knowledge Injection (SKI) Module

The Semantic Knowledge Injection (SKI) module explicitly incorporates **semantic knowledge** to assist visual anomaly detection.

First, large language models (e.g., ChatGPT and SparkDesk) are prompted to generate words related to:

- Normal scenes and actions.
- Anomaly-related scenes and concepts.

These words form a phrase list denoted as $M_{prior}$.

The phrase list is encoded by CLIP's text encoder to obtain semantic embeddings:

$$
F_{text}=\Phi_{CLIP-t}(\Phi_{token}(M_{prior}))
$$

where:

- $F_{text}$ are the textual embeddings.
- $\Phi_{CLIP-t}$ is CLIP's text encoder.
- $\Phi_{token}$ is the tokenizer.

To incorporate semantic knowledge into visual features, the authors propose a **cross-modal injection strategy**:

$$
F_{know}=\operatorname{sigmoid}(x_tF_{text}^{\top})\frac{F_{text}}{l}
$$

where $F_{know}$ represents the injected semantic knowledge.

The authors use **sigmoid** instead of **softmax** so that visual features can be associated with multiple semantic concepts.

Finally, $F_{know}$ is concatenated with the visual features $x_t$, and the resulting representation is fed into a binary detector to produce the **frame-level anomaly confidence** for the class-agnostic detection task.

## 3.4 Novel Anomaly Synthesis (NAS) Module

The Novel Anomaly Synthesis (NAS) module aims to improve the model's generalization to **novel anomaly categories** by generating **pseudo anomaly samples**.

The process consists of three steps:

1. **Generate synthetic anomaly content:** LLMs produce textual descriptions of potential novel anomalies, which are then converted into images or short videos using AIGC models.
2. **Create video clips:** Generated images are transformed into short video clips using a simple animation strategy.
3. **Pseudo anomaly synthesis:** The generated clips are randomly inserted into normal videos to create synthetic anomaly videos.

The resulting pseudo anomaly samples are used to **fine-tune** the model, improving its ability to detect and categorize unseen anomalies.

## 3.5 Objective Functions

During the first training stage (without pseudo anomaly samples), the model is optimized using three loss functions corresponding to its different components.

---

### 1. Class-Agnostic Detection Loss ($L_{bce}$)

This branch predicts whether a video contains an anomaly.

Following previous WSVAD methods, the model uses the **Top-K mechanism**:

- For abnormal videos, the top $K=n/16$ frames with the highest anomaly confidence are selected.
- For normal videos, all frames are used ($K=n$).

The selected anomaly scores are averaged to obtain a video-level prediction, which is passed through a sigmoid function. The prediction is then compared with the binary video label using **Binary Cross Entropy (BCE)**.

---

### 2. Class-Specific Categorization Loss ($L_{ce}$)

This branch predicts the anomaly category.

Since only video-level labels are available, frame features are aggregated into a video representation using a **soft attention mechanism**:

$$
x_{agg}=softmax(p)^{\top}x_t
$$

where:

- $p$ is the frame-level anomaly confidence.
- $x_t$ are the frame-level features.

The aggregated video feature is compared with the textual category embeddings, and the classification loss is computed using **Cross Entropy**.

Additionally, the textual category embeddings are enhanced by appending **learnable prompts**, following CoOp.

---

### 3. Semantic Similarity Loss ($L_{sim}$)

This loss explicitly optimizes the semantic knowledge embeddings introduced by the SKI module.

For **normal videos**, visual features are encouraged to have:

- High similarity with normal semantic knowledge.
- Low similarity with abnormal semantic knowledge.

This produces the loss:

$$
L_{sim-n}
$$

For **abnormal videos**, visual features are encouraged to have high similarity with abnormal semantic knowledge.

Since frame-level annotations are unavailable, the model uses the **Top-K mechanism** to locate abnormal regions before computing:

$$
L_{sim-a}
$$

The total semantic similarity loss is:

$$
L_{sim}=L_{sim-n}+L_{sim-a}
$$

---

### Overall Training Loss

The final objective function is the sum of the three losses:

$$
L_{train}=L_{bce}+L_{ce}+L_{sim}
$$

where:

- $L_{bce}$ optimizes anomaly detection.
- $L_{ce}$ optimizes anomaly categorization.
- $L_{sim}$ aligns visual features with semantic knowledge.

## 3.5.2 Fine-tuning Stage with Pseudo Anomaly Samples

After generating the pseudo anomaly samples $V_{nas}$ with the NAS module, the model is fine-tuned.

Since $V_{nas}$ provides **frame-level annotations**, the model can be trained with **full supervision** for anomaly detection.

For anomaly categorization, the classification loss remains the same:

$$
L_{ce2}=L_{ce}
$$

The difference is that the labels now include both **base** and **novel** anomaly categories.

For anomaly detection, the model uses a **frame-level Binary Cross Entropy** loss:

$$
L_{bce2}
$$

### Overall Fine-tuning Loss

The final loss during the fine-tuning stage is:

$$
L_{tune}=L_{bce2}+L_{ce2}+\lambda(L_{bce}+L_{ce})
$$

where:

- $L_{bce2}$ is the frame-level detection loss.
- $L_{ce2}$ is the anomaly categorization loss.
- $\lambda(L_{bce}+L_{ce})$ preserves the original weakly supervised objectives during fine-tuning.

## 4.2 Comparison with State-of-the-Arts

The authors compare their method with existing VAD approaches on three public benchmarks.

Since previous methods are designed for **closed-set VAD**, the comparison mainly focuses on **open-set detection**.

For a fair comparison, most competing methods are re-implemented using the same visual features as the proposed approach.

Methods marked with **†** are trained using the **entire training set**, including novel anomaly samples, giving them an advantage over methods trained without novel anomalies.

Despite this, the proposed method achieves performance comparable to or better than the best existing approaches.

Compared with the best-performing model (DMU), the proposed method improves:

- **+1.26% AUC** on UCF-Crime.
- **+2.63% AP** on XD-Violence.
- **+3.03% AUC** on UBnormal.

The proposed method shows a particularly clear advantage on **novel anomaly categories**.

The authors also compare their approach with Zhu et al. [67], the first work on Open-Set VAD, and report superior performance on both the UCF-Crime and XD-Violence datasets.

## OVVAD Overview

OVVAD is built upon **CLIP** and consists of two complementary branches.

### 1. Detection Branch

The video is processed by CLIP's image encoder to extract **frame-level visual features**.

These features are refined by the **Temporal Adapter (TA)** and **Semantic Knowledge Injection (SKI)** modules before being fed into a binary detector, which produces a **frame-level anomaly confidence score**.

```text
Video
   ↓
CLIP Image Encoder
   ↓
Frame Features
   ↓
TA + SKI
   ↓
Binary Detector
   ↓
Frame-level Anomaly Score
```

---

### 2. Categorization Branch

The frame-level features are aggregated into a **video-level representation**.

Meanwhile, anomaly category names are encoded by CLIP's text encoder to obtain **text embeddings**.

Finally, the model predicts the anomaly category by aligning the **video embedding** with the **text embeddings** using a cross-modal alignment mechanism.

```text
Video Embedding
        │
        │  Cross-modal Alignment
        │
Text Embeddings (CLIP Text Encoder)
        │
        ↓
Semantic Category
```

In summary, OVVAD uses CLIP's **image encoder** for anomaly detection and CLIP's **text encoder** to semantically categorize detected anomalies through visual-text alignment.

## Comparison: Traditional WSVAD vs. OVVAD

| Traditional WSVAD | OVVAD |
|-------------------|-------|
| Focuses on **anomaly detection**. | Focuses on **anomaly detection and semantic categorization**. |
| Outputs a frame-level anomaly score. | Outputs a frame-level anomaly score and an anomaly category. |
| Uses visual features for binary classification (normal vs. abnormal). | Uses visual features for detection and visual-text alignment for categorization. |
| Can sometimes detect unseen anomalies, but only labels them as **"anomaly"**. | Detects both seen and unseen anomalies and assigns them a **semantic label**. |
| Closed-set or Open-set detection. | Open-Vocabulary detection and categorization. |

### Pipeline Comparison

**Traditional WSVAD**

```text
Video
   ↓
Visual Features
   ↓
Detector
   ↓
Frame-level Anomaly Score
```

**OVVAD**

```text
                Video
                  │
          Visual Features
                  │
        ┌─────────┴─────────┐
        │                   │
 Detection Branch    Categorization Branch
        │                   │
Frame-level Score    CLIP Text Alignment
        │                   │
        └─────────┬─────────┘
                  │
      Semantic Anomaly Category
```

### Key Difference

The main contribution of OVVAD is the **categorization branch**, which aligns video features with CLIP text embeddings to assign semantic labels to both **seen** and **unseen** anomalies.
