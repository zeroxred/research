Algo interesante

Hay un detalle importante:

ninguno de estos papers pertenece a la línea de Weakly Supervised VAD (MIL) que tú has estado estudiando.

La mayoría siguen el paradigma clásico de Unsupervised VAD, es decir:

```text
Video
   ↓
Encoder
   ↓
Mamba
   ↓
Prediction / Reconstruction
   ↓
Prediction Error
   ↓
Anomaly Score
```

Su objetivo es aprender el patrón de normalidad y detectar desviaciones mediante errores de reconstrucción o predicción.

Mientras que tu línea de investigación es diferente

Tú has seguido esta evolución:

```text
MIL (Sultani)
      ↓
RTFM
      ↓
VadCLIP
      ↓
OVVAD
```

que corresponde a Weakly Supervised VAD, donde existen videos etiquetados como normales o anómalos y se busca localizar las anomalías mediante aprendizaje débil.

Eso abre una oportunidad interesante

Hasta donde llega la literatura reciente, no hay una línea consolidada que combine Mamba con los modelos modernos de WSVAD basados en CLIP, como VadCLIP u OVVAD. Los trabajos con Mamba se concentran sobre todo en el paradigma no supervisado.

Eso significa que una idea como:

```text
Frames
      ↓
CLIP
      ↓
Mamba
      ↓
MIL
      ↓
Anomaly Score
```

o incluso

```text
Frames
      ↓
CLIP
      ↓
Selective State Space
      ↓
Open-Vocabulary Alignment
      ↓
Anomaly Category
```

no corresponde a una dirección ya saturada por la literatura actual. Habría que revisar cuidadosamente los trabajos más recientes, pero a día de hoy parece ser una combinación con bastante espacio para explorar.
