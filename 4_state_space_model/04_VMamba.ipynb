### VMamba (Vision Mamba)

VMamba extends Mamba from **1D sequential modeling to 2D image understanding** by introducing **2D Selective Scan (SS2D)**. Instead of processing image patches as a single serialized sequence (as in Vision Mamba), VMamba performs **four directional scans** (left→right, right→left, top→bottom, and bottom→top). Each scan is processed independently by a Mamba (S6) block, and the outputs are merged to reconstruct the final 2D feature map.

By exploiting the inherent **2D spatial structure** of images, VMamba captures spatial dependencies more effectively while maintaining Mamba's **linear computational complexity**. This design improves visual representation learning and benefits both image classification and dense vision tasks such as object detection and semantic segmentation.

| Aspect | Classical SSM (S4) | Mamba (Selective SSM) | VMamba |
|--------|---------------------|------------------------|---------|
| **Input** | 1D sequence | 1D sequence | 2D image |
| **State Update** | $h_t=\bar{A}h_{t-1}+\bar{B}x_t$ | $h_t=\bar{A}_th_{t-1}+\bar{B}_tx_t$ | Same as Mamba |
| **Output** | $y_t=C h_t$ | $y_t=C_th_t$ | Same as Mamba |
| **State Matrix ($A$)** | Fixed | Fixed | Fixed |
| **Input Matrix ($B$)** | Fixed | $B_t=s_B(x_t)$ | Same |
| **Output Matrix ($C$)** | Fixed | $C_t=s_C(x_t)$ | Same |
| **Step Size ($\Delta$)** | Fixed | $\Delta_t=\tau(\theta_\Delta+s_\Delta(x_t))$ | Same |
| **Selection** | None | Dynamic parameter selection using $s_B(x)$, $s_C(x)$ and $s_\Delta(x)$ | Same dynamic selection |
| **Scan Operation** | Standard sequential state update | **Selective Scan** executes the dynamic SSM using token-dependent parameters | Same Selective Scan |
| **Spatial Processing** | Not supported | Image must be serialized into one 1D sequence | **SS2D:** four directional scans (← → ↑ ↓) using Cross-Scan and Cross-Merge |
| **Main Contribution** | Linear Time-Invariant (LTI) State Space Model | Introduces **input-dependent (selective) state updates** while preserving linear complexity | Extends Mamba to images by introducing **2D Selective Scan (SS2D)** |
| **Complexity** | Linear | Linear | Linear |
| **Intuition** | Every token is processed using the same dynamics. | Each token dynamically decides how information is **written ($B$)**, **propagated ($\Delta$)** and **read ($C$)** from the hidden state. | Instead of scanning a single sequence, the image is transformed into four directional sequences so the same Selective Scan can efficiently model 2D spatial dependencies. |
