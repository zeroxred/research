# Prediction and Reconstruction based VAD

## Reconstruction

Consists of reconstructing the input frame (or its representation). For example, if the input is a frame containing people walking, the model takes that frame as input and tries to reconstruct a similar version of the same frame:

$$
F \approx F'
$$

This can be useful for VAD because the model can be trained primarily on normal data. For example, consider a sequence of frames:

`[walking, walking, walking]`

During training, the model learns to reconstruct patterns that are present in normal videos. When it receives a normal frame, such as a frame showing people walking, it should be able to reconstruct it with relatively low error.

If an anomalous pattern appears, for example:

`[walking, running, walking]`

the model may reconstruct the anomalous frame less accurately because the anomalous pattern is not well represented in the normal training data. Therefore, the reconstruction error for that frame may be higher:

$$
E(F,F') \uparrow
$$

A high reconstruction error can then be used as an anomaly score, where larger errors indicate a higher likelihood that the frame contains anomalous content.
