# Bel and Decibel (dB): A Quick Tutorial

## 1. What is a Bel (B)?

A **bel (B)** is a logarithmic unit used to express the ratio between two power levels.

$$
L_{\mathrm{B}}=\log_{10}\left(\frac{P_2}{P_1}\right)
$$

Where:
- $P_2$: measured power.
- $P_1$: reference power.
- $L_{\mathrm{B}}$: power-level difference in bels.

### Understanding the logarithmic scale

The logarithm tells us which power of 10 produces a given ratio.

| Power ratio ($P_2/P_1$) | Logarithmic calculation | Difference | Interpretation |
|---:|---|---:|---|
| 1 | $\log_{10}(1)=0$ | 0 B | Same power |
| 10 | $\log_{10}(10)=1$ | 1 B | 10 times more power |
| 100 | $\log_{10}(100)=2$ | 2 B | 100 times more power |
| 1,000 | $\log_{10}(1000)=3$ | 3 B | 1,000 times more power |
| 0.1 | $\log_{10}(0.1)=-1$ | −1 B | One-tenth of the reference power |
| 0.01 | $\log_{10}(0.01)=-2$ | −2 B | One-hundredth of the reference power |

**Example:** If the reference power is 1 W and the measured power is 100 W:

$$
L_{\mathrm{B}}=\log_{10}\left(\frac{100}{1}\right)=2\ \mathrm{B}
$$

Therefore, the measured power is **2 B greater than the reference, meaning 100 times as much power**.

## 2. What is a Decibel?

A **decibel (dB)** is one-tenth of a bel:

$$
1\ \mathrm{B}=10\ \mathrm{dB}
$$

For power ratios:

$$
\boxed{L_{\mathrm{dB}}=10\log_{10}\left(\frac{P_2}{P_1}\right)}
$$

For the example:

$$
10\log_{10}(100/10)=10\ \mathrm{dB}
$$

## 3. Understanding Decibels

| Power ratio | Level difference |
|---:|---:|
| 0.1 | −10 dB |
| 0.5 | −3.01 dB |
| 1 | 0 dB |
| 2 | +3.01 dB |
| 10 | +10 dB |
| 100 | +20 dB |

**Key points:**
- Positive dB: the measured power is greater than the reference.
- Negative dB: the measured power is lower than the reference.
- Zero dB: both powers are equal; this does not mean zero power.

## 4. Power vs. Amplitude

For power ratios, use:

$$
L_{\mathrm{dB}}=10\log_{10}\left(\frac{P_2}{P_1}\right)
$$

For amplitude ratios, use:

$$
L_{\mathrm{dB}}=20\log_{10}\left(\frac{A_2}{A_1}\right)
$$

The factor 20 applies when power is proportional to the square of amplitude and the relevant conditions are satisfied (e.g., equal impedance for voltage comparisons).

## 6. Practice

A signal has a power of 50 W, while the reference signal has a power of 5 W. What is the difference in decibels?

$$
L_{\mathrm{dB}}=10\log_{10}(50/5)
$$

**Answer: 10 dB**, because the power ratio is 10.

# Signal-to-Noise Ratio

Ratio of signal power to noise power, often used in db.

- **SNR high:** signal high noise lower
- **SNR lower:** signal lower noise high

$$
SNR=\frac{P_{\text{signal}}}{P_{\text{noise}}}
$$

can be also be expressed in dB:

$$
\mathrm{SNR}_{\mathrm{dB}}=10\log_{10}\left(\frac{P_{\text{signal}}}{P_{\text{noise}}}\right)
$$

Where:

- $P_{\text{signal}}$: Power of the information of interest.
- $P_{\text{noise}}$: Power of interference or unwanted information.
- $\log_{10}$: Logarithm to base 10.

Examples:

| SNR (dB) | Interpretation |
|---:|---|
| $-10$ dB | The noise has 10 times more power than the signal. |
| $0$ dB | The signal and noise have the same power. |
| $10$ dB | The signal has 10 times more power than the noise. |
| $20$ dB | The signal has 100 times more power than the noise. |
| $30$ dB | The signal has 1,000 times more power than the noise. |

A negative SNR does not mean that the signal has disappeared; it means that its power is lower than the noise power.

## Peak Signal-to-Noise Ratio (PSNR)

The **Peak Signal-to-Noise Ratio (PSNR)** is a metric used to measure the quality of a reconstructed or processed signal, particularly images and videos, by comparing it with an original reference signal.

A higher PSNR generally indicates that the reconstructed image or video is more similar to the original, while a lower PSNR indicates greater reconstruction error.

### Formula

$$
\mathrm{PSNR}=10\log_{10}\left(\frac{MAX_I^2}{\mathrm{MSE}}\right)
$$

Where:

- $MAX_I$: Maximum possible pixel value in the image. For an 8-bit image, $MAX_I=255$.
- $\mathrm{MSE}$: Mean Squared Error between the original and reconstructed images.
- $\log_{10}$: Logarithm to base 10.

The Mean Squared Error (MSE) is calculated as:

$$
\mathrm{MSE}=\frac{1}{MN}\sum_{i=1}^{M}\sum_{j=1}^{N}\left(I(i,j)-\hat{I}(i,j)\right)^2
$$

Where:

- $M$ and $N$: Image height and width, respectively.
- $I(i,j)$: Pixel value at position $(i,j)$ in the original image.
- $\hat{I}(i,j)$: Pixel value at position $(i,j)$ in the reconstructed image.

### Example

| PSNR | Interpretation |
|---:|---|
| High PSNR | Low reconstruction error; the reconstructed image is generally more similar to the original. |
| Low PSNR | High reconstruction error; the reconstructed image differs more from the original. |
| Infinite PSNR | The original and reconstructed images are identical, resulting in zero MSE. |

PSNR is expressed in decibels (dB). Its value depends on the image's intensity range and the reconstruction error, so comparisons are most meaningful when the same data range and evaluation procedure are used.

### Application to Video Anomaly Detection

In video anomaly detection, PSNR can be used to evaluate how accurately a model reconstructs or predicts video frames.

The underlying idea is that a model trained primarily on normal patterns may reconstruct or predict normal frames more accurately than anomalous frames. Anomalous frames may therefore produce larger reconstruction or prediction errors and lower PSNR values.

However, this relationship is not guaranteed: some anomalies may still be reconstructed accurately, and normal frames may produce large errors due to motion, illumination changes, or other variations.

Consequently, PSNR can serve as an anomaly-related score, but it is not itself a direct measure of anomaly probability.

## Why Does PSNR Use the Maximum Pixel Value Squared?

The Peak Signal-to-Noise Ratio (PSNR) is defined as:

$$
PSNR=10\log_{10}\left(\frac{MAX_I^2}{MSE}\right)
$$

There are two important questions: **Why is the maximum pixel value squared, and why is it constant?**

### 1. Why Is the Maximum Pixel Value Squared?

The Mean Squared Error (MSE) is calculated by squaring the difference between each pixel in the original and reconstructed images:

$$
MSE=\frac{1}{N}\sum_{i=1}^{N}(I_i-\hat{I}_i)^2
$$

Where:

- $I_i$: Pixel value in the original image.
- $\hat{I}_i$: Pixel value in the reconstructed image.
- $N$: Total number of pixels.

Because the pixel differences are squared, the MSE is expressed in units of intensity squared.

The numerator, $MAX_I^2$, is also expressed in units of intensity squared. This makes the ratio dimensionless.

However, **using the maximum pixel value is a convention that defines PSNR; it is not an unavoidable consequence of the conventional SNR formula.**

### 2. Why Is the Maximum Pixel Value Constant?

Consider an 8-bit image. Each pixel can have an intensity value between 0 and 255:

- Black: 0
- Gray: 128
- White: 255

Therefore:

$$
MAX_I=255
$$

And:

$$
MAX_I^2=255^2=65025
$$

This value is constant because it depends on the image's permitted intensity range, not on the actual pixel values in a particular image.

For example, a dark image and a bright image can both use the same 8-bit representation. Their maximum possible pixel value remains 255.

In contrast, the MSE changes depending on how much the reconstructed image differs from the original.

### 3. A Numerical Example

Suppose we evaluate two reconstructions of 8-bit images.

**Reconstruction A: Low reconstruction error**

$$
PSNR=10\log_{10}\left(\frac{65025}{10}\right)
$$

$$
PSNR\approx38.13\text{ dB}
$$

**Reconstruction B: High reconstruction error**

$$
PSNR=10\log_{10}\left(\frac{65025}{1000}\right)
$$

$$
PSNR\approx18.13\text{ dB}
$$

The numerator remains constant because both images use the same 8-bit intensity range. The MSE changes according to the reconstruction error.

Consequently, reconstruction A has a higher PSNR because its error is smaller.

### 4. Why Must the Numerator Be a Scalar?

The MSE is a scalar that summarizes the reconstruction error across all pixels. Therefore, the numerator must also be a scalar to calculate the ratio.

However, **the fact that MSE is a scalar does not mean that we must use the maximum pixel value as the numerator**. We could use other scalar references, such as the actual average power of the original image.

The maximum is used because it is the reference chosen by the conventional PSNR definition, not because it is impossible to calculate the actual power of an image.

### 5. Could We Use the Actual Image Power Instead?

Yes. The average power of the original image can be calculated as:

$$
P_{\text{image}}=\frac{1}{N}\sum_{i=1}^{N}I_i^2
$$

We could then define a signal-to-noise ratio using the image's actual power:

$$
SNR=10\log_{10}\left(\frac{P_{\text{image}}}{MSE}\right)
$$

Unlike PSNR, this formulation uses the actual average image power rather than the maximum possible intensity squared.

These metrics use different reference values and therefore measure the error relative to different baselines.

### Key Takeaway

The numerator must be a scalar to calculate the ratio, but it does not necessarily have to be the maximum pixel value. **PSNR uses the maximum because of its specific mathematical definition, not because MSE is a scalar or because calculating the actual image power is impossible.**

The maximum remains constant because it represents the upper limit of the pixel intensity range, not the actual pixel values of an individual image.

```python
import numpy as np

def calculate_psnr(original, reconstructed, max_value=255.0):
    original = np.asarray(original, dtype=np.float64)
    reconstructed = np.asarray(reconstructed, dtype=np.float64)

    if original.shape != reconstructed.shape:
        raise ValueError("Las imágenes deben tener la misma forma.")

    mse = np.mean((original - reconstructed) ** 2)

    if mse == 0:
        return float("inf")

    return 10 * np.log10((max_value ** 2) / mse)


original = np.array([[100, 150], [200, 250]])
reconstructed = np.array([[100, 140], [190, 250]])

print("PSNR:", calculate_psnr(original, reconstructed), "dB")
```
