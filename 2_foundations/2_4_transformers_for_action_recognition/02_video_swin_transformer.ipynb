# Video Swin Transformer


Video Swin Transformer extends Swin Transformer from images to videos by introducing a temporal dimension in the patch embedding and window-based self-attention mechanisms, enabling efficient spatio-temporal feature learning while preserving the hierarchical architecture of the original Swin Transformer.

The model is basically the Swin Transformer, but with the time dimesion $T$. In this arquitecture the information is taken by two conseutives frames, so each patch has $2 \times 4 \times 4 \times 3 = 96$ values, hence the number of patches is $\frac{T}{2} \times \frac{H}{4} \times \frac{W}{4}$.

 A linear embedding layer is then applied to project the features of each token to an arbitrary dimension denoted by C.

## Video Swin Transformer Block

The major component of the architecture is the Video Swin Transformer block, which is built by replacing the multihead self-attention (MSA) module in the standard Transformer layer with the 3D shifted window based multi-head self-attention module and keeping the other components unchanged.
