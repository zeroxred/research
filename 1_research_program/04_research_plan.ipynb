Research plan 1

Objectives
The goal of this research is to design and implement deep learning models for visual information understanding, with a particular focus on:
-	Real-time object detection using models such as YOLO and SSD.
-	Scene and activity recognition in video sequences for security and behavioral analysis.
-	Text extraction (OCR) and semantic interpretation from images and scanned documents.
-	(Optional extension) Integration of language models (e.g., Transformers) to develop multimodal AI systems.

Specific Objectives
To achieve the goals outlined above, the following specific objectives are proposed:
-	To collect and prepare datasets for training models in image and video analysis.
-	To design and train real-time object detection models (e.g., YOLO or SSD) applicable to surveillance and public safety scenarios.
-	To implement OCR pipelines for text extraction from images and scanned documents, including preprocessing, classification, and semantic organization of the extracted content.
-	To develop an activity recognition system capable of detecting patterns and behaviors from video data.
-	To explore the integration of language models (e.g., Transformers) for tasks where textual information plays a central role, such as semantic image annotation or document interpretation.

RESEARCH PLAN 2

1. Background and Objective

Video surveillance is an important tool for public safety; however, manual monitoring is limited in efficiency and scalability. Therefore, automatic video analysis using computer vision techniques has become increasingly important, especially for detecting dangerous or abnormal human behaviors. This research aims to develop a video surveillance system based on spatiotemporal deep learning models to detect violent actions and anomalies with improved accuracy and robustness.

2. Methodology and Research Plan

The proposed approach combines object detection, multi-object tracking, and temporal modeling methods such as 3D Conv Nets and Transformer-based architectures. By integrating spatial and temporal information, the system aims to capture both individual actions and interactions among multiple subjects. First, supervised action recognition will be applied to classify scenarios such as violent versus normal activities. Then, anomaly detection methods will be introduced to identify unseen or unexpected behaviors using unsupervised or weakly supervised approaches.

The research will be conducted in the following stages:

- Implementation and evaluation of baseline action recognition models
- Integration of detection and tracking for multi-person scene analysis
- Exploration of anomaly detection methods and comparison of different model architectures
- Development of improved models to enhance generalization, robustness, and adaptability across environments

3. Expected Results

The expected outcome is a system capable of detecting critical events in video data, contributing to surveillance applications, advancements in computer vision, and improved public safety.
