## Unsupervised Video Anomaly Detection (VAD)

Since unsupervised VAD datasets contain only **normal events**, the model cannot be trained directly to distinguish normal from anomalous behavior. Instead, these methods rely on a **proxy task**: an auxiliary learning objective that forces the model to learn the distribution of normal video patterns.

### Proxy Tasks

The two dominant proxy tasks are:

- **Reproduction-based methods (Reconstruction in the literature):**  The model, typically implemented as an **Autoencoder**, receives a video frame (or its representation) and learns to reproduce the same frame as accurately as possible. Since it is trained only on normal data, it struggles to reproduce anomalous scenes, resulting in a high reproduction error.

	For example, if the input is a frame containing people walking, the model takes that frame as input and tries to reconstruct a similar version of the same frame:

	$
		\begin{equation}
		F \approx F'
		\end{equation}
	$

	This can be useful for VAD because the model can be trained primarily on normal data. For example, consider a sequence of frames:

	`[walking, walking, walking]`

	During training, the model learns to reconstruct patterns that are present in normal videos. When it receives a normal frame, such as a frame showing people walking, it should be able to reconstruct it with relatively low error.

	If an anomalous pattern appears, for example:

	`[walking, running, walking]`

	the model may reconstruct the anomalous frame less accurately because the anomalous pattern is not well represented in the normal training data. Therefore, the reconstruction error for that frame may be higher:

	$$
	E(F,F') \uparrow
	$$

	A high reconstruction error can then be used as an anomaly score, where larger errors indicate a higher likelihood that the frame contains anomalous content.


- **Prediction-based methods:**  
  The model receives a sequence of previous frames and predicts the next frame. This requires learning not only the spatial appearance of the scene but also its temporal dynamics. When an abnormal event occurs, the prediction error increases because the observed motion deviates from the learned normal behavior.

Although reconstruction-based methods were introduced first, prediction-based approaches have become the dominant paradigm because they explicitly model temporal information, which is essential for detecting anomalies in videos.
