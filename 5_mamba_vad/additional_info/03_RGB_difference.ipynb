## RGB Differences Between Adjacent Video Frames in STNMamba

STNMamba calculates the differences between the raw RGB values of consecutive video frames to capture temporal changes and help represent motion.

### 1. What do the terms mean?

- **Raw RGB values:** The original pixel values in the Red (R), Green (G), and Blue (B) channels.
- **Adjacent video frames:** Two consecutive frames in a video, such as frame *t* and frame *t+1*.
- **RGB differences:** The differences between the RGB values of pixels at the same spatial location in the two frames.

### 2. Illustrative example

Suppose a pixel has the following RGB values in two consecutive frames:

| Frame | Color | RGB Values |
|---|---|---|
| Frame *t* | 🟥 Red | (100, 50, 20) |
| Frame *t+1* | 🟧 Orange | (130, 60, 25) |

The RGB difference is calculated channel by channel:

$$
D_t(x,y)=I_{t+1}(x,y)-I_t(x,y)
$$

Therefore:

$$
D_t(x,y)=(130-100,\ 60-50,\ 25-20)
$$

**Result:** $(30, 10, 5)$

This result indicates how much each color channel changed between the two frames at that pixel location.

### 3. Why is this useful in STNMamba?

RGB differences provide information about temporal changes between consecutive frames. These changes may be caused by object motion, camera movement, or illumination changes.

- **Original RGB frames:** Represent the visual appearance of objects and scenes.
- **RGB difference frames:** Represent changes in pixel intensities between consecutive frames.

This information can help the model learn temporal patterns that are useful for video anomaly detection.

### 4. RGB differences vs. optical flow

RGB differences are not the same as optical flow.

- **RGB differences** measure changes in pixel intensity and color.
- **Optical flow** estimates the direction and magnitude of apparent pixel motion between frames.

RGB differences are computationally simpler, but they do not directly estimate motion direction.

**In summary:** STNMamba uses RGB differences between consecutive video frames to provide temporal information without requiring the additional computation of optical flow.
