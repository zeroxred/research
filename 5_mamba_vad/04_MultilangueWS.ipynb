# Multilingual-Prompt-Guided Directional Feature Learning for Weakly Supervised Video Anomaly Detection (TPAMI 2025)

## Context

Weakly Supervised Video Anomaly Detection (WSVAD) trains using **video-level labels** instead of frame-level annotations.

```text
Normal Video     → Normal

Abnormal Video   → Abnormal
```

The model must automatically discover **which temporal segments are anomalous**.

Recent methods such as **VadCLIP** leverage Vision-Language Models (CLIP) to align visual features with textual descriptions of normal and abnormal events.

---

# Motivation

Prompt-based methods have significantly improved WSVAD, but they still present several limitations.

## 1. Prompt ambiguity

Using a single prompt to represent an anomaly is often insufficient.

For example, the same event may be described as:

- Fighting
- Violence
- Assault
- Physical attack

Although semantically similar, these prompts generate different text embeddings.

---

## 2. Language limitation

Most existing methods only use English prompts.

However, CLIP has been trained with multilingual data, meaning that different languages may provide complementary semantic information.

For example:

```text
Fight

Pelea

打架

Combat
```

Each language captures slightly different semantic characteristics.

---

## 3. Temporal modeling

Vision-Language Models mainly learn cross-modal alignment but are not specifically designed to model long temporal dependencies.

Therefore, additional temporal modeling is required to accurately localize anomalous segments.

---

# Main Idea

The paper proposes three complementary improvements.

## 1. Multilingual Prompt Learning

Instead of describing "normal" and "anomaly" using only English, the model constructs prompt pools in multiple languages.

```text
Normal

English
Chinese
...

Anomaly

English
Chinese
...
```

The objective is to enrich the semantic representation of both concepts.

The authors argue that multilingual prompts provide complementary semantic information that cannot be obtained from a single language.

---

## 2. Adaptive Top-K Prompt Selection

Rather than using every prompt equally, the model dynamically selects the most relevant prompts for each video snippet.

```text
Prompt Pool

↓

Top-K Selection

↓

Most relevant prompts
```

This prevents irrelevant prompts from introducing noisy supervision.

---

## 3. Multi-Granularity Temporal Modeling

To improve visual feature learning, the paper combines **Transformer** and **Mamba**.

Unlike replacing Transformer entirely, both modules play complementary roles.

### Mamba

Responsible for modeling long-range temporal dependencies across coarse video snippets.

```text
Coarse Snippets

↓

Mamba

↓

Global temporal relationships
```

---

### Transformer

Uses the contextual information generated by Mamba to refine local temporal details.

```text
Fine Snippets

↓

Transformer

↓

Fine-grained refinement
```

Therefore, the architecture exploits both:

- long-range temporal reasoning (Mamba)
- local temporal refinement (Transformer)

---

# Directional Feature Learning

Besides the multilingual prompt supervision, the paper introduces a **Gradual Directional Loss**.

Instead of only separating normal and abnormal features, the optimization progressively encourages:

- normal features to become closer;
- abnormal features to become closer;
- normal and abnormal distributions to move farther apart.

The optimization therefore learns a more discriminative feature space.

---

# Overall Pipeline

```text
Video

↓

Feature Extraction

↓

Multi-Granularity Module
(Mamba + Transformer)

↓

Visual Features

↓

Similarity with
Multilingual Prompt Pools

↓

Adaptive Top-K Prompt Selection

↓

Directional Feature Learning

↓

Anomaly Score
```

---

# Main Contributions

- Introduces multilingual prompts for WSVAD.
- Uses multiple prompts instead of a single textual description.
- Dynamically selects the most relevant prompts through Top-K selection.
- Combines Mamba and Transformer for complementary temporal modeling.
- Proposes a Gradual Directional Loss to improve feature discrimination.
- Achieves state-of-the-art performance on four WSVAD benchmarks while also demonstrating good generalization to medical time-series (EMG and ECG).

---

# Relationship with VadCLIP

Conceptually, this work can be viewed as an evolution of VadCLIP.

| VadCLIP | This Paper |
|----------|------------|
| CLIP | CLIP |
| Prompt Engineering | Multilingual Prompt Engineering |
| Single Prompt | Multiple Prompts |
| Fixed Prompt Usage | Adaptive Top-K Prompt Selection |
| Temporal Modeling | Transformer + Mamba |
| Cross-modal Alignment | Cross-modal Alignment + Directional Feature Learning |

Unlike VAD papers such as STNMamba or VADMamba++, this work **does not introduce a new anomaly detection paradigm**.

Instead, it refines the existing Vision-Language framework by improving:

- semantic guidance through multilingual prompts;
- temporal feature learning through Transformer + Mamba;
- feature discrimination through directional optimization.
