# 🗺️ Research Roadmap

## Objective
Develop a real-time video surveillance system using spatiotemporal deep learning, progressing from action recognition to anomaly detection.

## Phase 0 — Foundations (3 months)

### Level 1 — Fundamentals

- **Learning Spatiotemporal Features with 3D Convolutional Networks**  
  Du Tran, Lubomir Bourdev, Rob Fergus, Lorenzo Torresani, Manohar Paluri (2015)

- **Two-Stream Convolutional Networks for Action Recognition in Videos**  
  Karen Simonyan, Andrew Zisserman (2014)

  **Note: [Conceptual/Theoretical]** These works will be studied to understand the transition from 2D to spatiotemporal modeling and the importance of temporal cues such as optical flow. They will serve as theoretical background rather than implementation targets.

### Level 2 — Advanced CNN-based Models

- **SlowFast Networks for Video Recognition**  
  Christoph Feichtenhofer, Haoqi Fan, Jitendra Malik, Kaiming He (2019)

  **Note: [Practical/Black-Box]** This model will be analyzed to understand multi-scale temporal processing. It will be used as a feature extractor with pre-trained weights, and its input-output representations will be examined for integration into the pipeline.

### Level 3 — Transformer-based Models

- **Is Space-Time Attention All You Need for Video Understanding?**  
  Gedas Bertasius, Heng Wang, Lorenzo Torresani (2021)

- **Video Swin Transformer**  
  Ze Liu, Jia Ning, Yue Cao, Yixuan Wei, Zheng Zhang, Stephen Lin, Baining Guo (2022)

  **Note: [Conceptual/Architectural]** The study will focus on the application of self-attention mechanisms to temporal modeling. Special attention will be given to computational complexity and its implications for real-time processing.

### Level 4 — Core Video Anomaly Detection

- **Real-World Anomaly Detection in Surveillance Videos**  
  Waqas Sultani, Chen Chen, Mubarak Shah (2018)

- **Weakly-Supervised Video Anomaly Detection with Robust Temporal Feature Magnitude Learning (RTFM, ICCV 2021)**  
  Yu Tian et al.

- **VadCLIP: Adapting Vision-Language Models for Weakly Supervised Video Anomaly Detection**  
  Peng Wu et al. (AAAI 2024)

- **Open-Vocabulary Video Anomaly Detection**  
  Peng Wu et al. (CVPR 2024)

**Note: [Core Research Direction]**  
This level focuses on understanding the evolution of Video Anomaly Detection (VAD), from early Multiple Instance Learning (MIL)-based formulations to modern Vision-Language approaches. Particular attention will be given to weakly-supervised learning, anomaly localization, feature representation, and the integration of foundation models such as CLIP. The objective is to understand how anomaly detection formulations have evolved and to identify limitations that remain relevant for real-world surveillance systems.

### Level 5 — Research Landscape & Emerging Paradigms (optional)

- **Generalized Video Anomaly Event Detection: Systematic Taxonomy and Comparison of Deep Models**  
  (2023)

- **Deep Learning for Video Anomaly Detection: A Review**  
  (2024)

- Recent works on:
  - Open-Set Video Anomaly Detection
  - Open-Vocabulary Video Anomaly Detection
  - Vision-Language Models for Video Understanding
  - Multimodal Video Anomaly Detection (2024–2025)

**Note: [Research Mapping]**  
This level aims to build a comprehensive view of the VAD research landscape. Survey papers will be used to identify major research directions, benchmark datasets, evaluation protocols, and unresolved challenges. The goal is to establish a broad understanding of the field before focusing on specific architectural innovations.

### Level 6 — Efficient Temporal Modeling

- **Mamba: Linear-Time Sequence Modeling with Selective State Spaces**  
  Albert Gu, Tri Dao (ICLR 2024 / arXiv 2023)

- **Vision Mamba: Efficient Visual Representation Learning with Bidirectional State Space Model**  
  Lianghui Zhu et al. (ICML 2024)

- **VideoMamba: State Space Model for Efficient Video Understanding**  
  Kunchang Li et al. (ECCV 2024)

- **VMamba: Visual State Space Model**  
  Yue Liu et al. (NeurIPS 2024)

**Note: [Architectural Exploration]**  
The focus of this level is to study alternatives to Transformer-based temporal modeling. Special attention will be given to State Space Models (SSMs), their computational properties, scalability, and suitability for long video sequences and real-time surveillance applications. The objective is to evaluate whether these architectures can provide a practical trade-off between performance and efficiency in VAD systems.

### Level 7 — Frontier Topics for Surveillance VAD

- Recent works on:
  - Unified Video Anomaly Detection
  - Multimodal VAD
  - Audio-Visual Anomaly Detection
  - Foundation Models for Surveillance
  - Continual Learning for VAD
  - Domain Adaptation in Surveillance Videos
  - Real-Time Edge Deployment

**Note: [Future Research Opportunities]**  
This level explores emerging research directions that may serve as potential thesis topics or future research projects. The emphasis is on robustness, generalization, deployment constraints, and adaptability to real-world surveillance environments.

---

### Goals
- Develop a solid understanding of spatiotemporal representations for video analysis
- Study and analyze key approaches in action recognition and anomaly detection to understand the current state of the art
- Identify the strengths and limitations of different modeling paradigms (e.g., CNN-based, Transformer-based, and emerging efficient architectures)
- Gain preliminary hands-on experience through limited and controlled experiments to support theoretical understanding
- Establish a clear research direction for subsequent phases, where more extensive implementations and experimentation may be conducted
- Develop a deep understanding of modern Video Anomaly Detection (VAD) methodologies for surveillance scenarios
- Analyze the evolution from Multiple Instance Learning (MIL)-based approaches to Vision-Language and Open-Vocabulary frameworks
- Understand benchmark datasets, evaluation protocols, and the limitations of current anomaly detection systems
- Explore emerging paradigms such as foundation models, multimodal learning, and open-set anomaly detection
- Study efficient temporal modeling architectures, including State Space Models (SSMs), for long-video and real-time processing
- Identify promising research opportunities that balance anomaly detection performance, computational efficiency, and deployment feasibility in real-world surveillance systems
- Establish a solid foundation for future experimentation, reproduction studies, and thesis-oriented research

## Phase 1 — Action Recognition (1-3 months)

### Datasets
- UCF-Crime

### Tasks
- Implement a baseline action recognition system:
  - Fight vs non-fight classification
- Experiment with:
  - 3D CNNs (C3D)
  - Transformer-based models

### Goals
- Build a complete pipeline:
  - video → model → prediction
- Establish baseline performance

---

## Phase 2 — Realistic Surveillance System (3-4 months)

### Add Components
- Object detection:
  - YOLO-based models
- Multi-object tracking:
  - ByteTrack / DeepSORT

### Datasets
- UCF-Crime (primary)
- XD-Violence (complementary)

### Tasks
- Extend to multi-class detection:
  - fight
  - robbery
- Integrate spatial + temporal modeling

### Goals
- Build a surveillance-like system
- Improve robustness and performance

---

## Phase 3 — Anomaly Detection (4-6 months)

### Topics to Study
- Multiple Instance Learning (MIL)
- Autoencoders for anomaly detection
- Transformer-based anomaly detection

### Datasets
- UCF-Crime
- ShanghaiTech
- Avenue

### Tasks
- Implement anomaly detection models
- Generate anomaly scores over time

### Goals
- Detect unseen behaviors
- Handle weak supervision scenarios

---

## Phase 4: Advanced Research and Model Development (6–12+ months)

### Focus Areas
- Hybrid models (Action Recognition + Anomaly Detection)
- Real-time optimization (low latency, efficient models)
- Cross-domain generalization
- Reduction of false positives

### Goal
- Develop a novel contribution suitable for publication
