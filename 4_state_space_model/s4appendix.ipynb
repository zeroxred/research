# Appendix A. From State-Space Recurrence to Convolution

After discretization, the continuous-time state-space model becomes

$$
x_{k+1}=\bar{A}x_k+\bar{B}u_k,
$$

$$
y_k=\bar{C}x_k.
$$

where:

- $u_k$ is the sampled input sequence.
- $x_k$ is the discrete-time state.
- $y_k$ is the discrete-time output.

For simplicity, assume the initial state is

$$
x_{-1}=0.
$$

---

## A.1 Unrolling the State Recurrence

The recurrence can be expanded explicitly by repeatedly substituting the previous state.

### Step 1

At the first time step,

$$
x_0=\bar{B}u_0.
$$

Therefore,

$$
y_0=\bar{C}x_0
=\bar{C}\bar{B}u_0.
$$

---

### Step 2

At the next time step,

$$
x_1=\bar{A}x_0+\bar{B}u_1.
$$

Substituting $x_0$,

$$
x_1
=
\bar{A}(\bar{B}u_0)
+
\bar{B}u_1,
$$

which simplifies to

$$
x_1
=
\bar{A}\bar{B}u_0
+
\bar{B}u_1.
$$

The corresponding output is

$$
y_1
=
\bar{C}x_1
=
\bar{C}\bar{A}\bar{B}u_0
+
\bar{C}\bar{B}u_1.
$$

Notice that the previous input $u_0$ still influences the system through the state-transition matrix $\bar{A}$.

---

### Step 3

Repeating the same procedure,

$$
x_2
=
\bar{A}x_1+\bar{B}u_2.
$$

Substituting $x_1$,

$$
x_2
=
\bar{A}
\left(
\bar{A}\bar{B}u_0
+
\bar{B}u_1
\right)
+
\bar{B}u_2.
$$

Distributing $\bar{A}$,

$$
x_2
=
\bar{A}^2\bar{B}u_0
+
\bar{A}\bar{B}u_1
+
\bar{B}u_2.
$$

Thus,

$$
y_2
=
\bar{C}\bar{A}^2\bar{B}u_0
+
\bar{C}\bar{A}\bar{B}u_1
+
\bar{C}\bar{B}u_2.
$$

---

## A.2 Identifying the Pattern

The first outputs are therefore

$$
y_0
=
\bar{C}\bar{B}u_0,
$$

$$
y_1
=
\bar{C}\bar{A}\bar{B}u_0
+
\bar{C}\bar{B}u_1,
$$

$$
y_2
=
\bar{C}\bar{A}^2\bar{B}u_0
+
\bar{C}\bar{A}\bar{B}u_1
+
\bar{C}\bar{B}u_2.
$$

A clear pattern emerges:

- Each past input contributes to the current output.
- Older inputs are multiplied by higher powers of the state-transition matrix $\bar{A}$.
- The influence of an input depends only on how many time steps have elapsed since it entered the system.

Physically, this reflects the memory of the dynamical system. An input affects the output immediately, and its influence continues over time through repeated applications of the transition matrix.

This pattern can be generalized as

$$
y_k
=
\bar{C}\bar{A}^k\bar{B}u_0
+
\bar{C}\bar{A}^{k-1}\bar{B}u_1
+\cdots+
\bar{C}\bar{A}\bar{B}u_{k-1}
+
\bar{C}\bar{B}u_k.
$$

This expression is obtained directly by repeatedly expanding the recurrence relation.

---

## A.3 Defining the SSM Kernel

Notice that the same coefficients appear repeatedly:

$$
\bar{C}\bar{B},
\quad
\bar{C}\bar{A}\bar{B},
\quad
\bar{C}\bar{A}^2\bar{B},
\quad
\ldots
$$

These coefficients depend only on the state-space matrices and are independent of the input sequence.

Therefore, we define the **SSM convolution kernel**

$$
\bar{K}
=
\left(
\bar{C}\bar{B},
\bar{C}\bar{A}\bar{B},
\bar{C}\bar{A}^2\bar{B},
\ldots
\right).
$$

Equivalently,

$$
K_i
=
\bar{C}\bar{A}^{i}\bar{B}.
$$

---

## A.4 Discrete Convolution

Before rewriting the state-space model, let us recall the definition of a discrete convolution.

Given an input sequence

$$
u=(u_0,u_1,\ldots,u_{L-1}),
$$

and a convolution kernel

$$
K=(K_0,K_1,\ldots,K_{L-1}),
$$

their discrete convolution is defined as

$$
y_k
=
\sum_{i=0}^{k}
K_i\,u_{k-i},
\qquad
k=0,\ldots,L-1.
$$

Expanding the first few outputs,

$$
y_0
=
K_0u_0,
$$

$$
y_1
=
K_1u_0
+
K_0u_1,
$$

$$
y_2
=
K_2u_0
+
K_1u_1
+
K_0u_2.
$$

A discrete convolution computes each output as the weighted sum of the current input and all previous inputs. The kernel acts as a set of weights that quantifies how much each past input contributes to the current output. As time advances, the kernel effectively slides over the input sequence, accumulating the influence of previous inputs.

This operation is fundamental in signal processing because it describes the input-output relationship of linear time-invariant (LTI) systems.

---

## A.5 From the State-Space Model to a Convolution

Now compare the convolution expansion with the expressions obtained by unrolling the state-space recurrence:

$$
y_0
=
\bar{C}\bar{B}u_0,
$$

$$
y_1
=
\bar{C}\bar{A}\bar{B}u_0
+
\bar{C}\bar{B}u_1,
$$

$$
y_2
=
\bar{C}\bar{A}^2\bar{B}u_0
+
\bar{C}\bar{A}\bar{B}u_1
+
\bar{C}\bar{B}u_2.
$$

Comparing these equations with the convolution expansion,

$$
y_0
=
K_0u_0,
$$

$$
y_1
=
K_1u_0
+
K_0u_1,
$$

$$
y_2
=
K_2u_0
+
K_1u_1
+
K_0u_2,
$$

reveals that they have exactly the same mathematical structure.

The correspondence is immediate:

$$
K_0=\bar{C}\bar{B},
$$

$$
K_1=\bar{C}\bar{A}\bar{B},
$$

$$
K_2=\bar{C}\bar{A}^2\bar{B},
$$

or, in general,

$$
K_i=\bar{C}\bar{A}^{i}\bar{B}.
$$

Consequently, the output sequence can be written compactly as

$$
y=\bar{K}*u.
$$

The convolution formulation is **not a different model**. It is exactly the same linear dynamical system expressed in a different mathematical form. Instead of recursively computing the hidden state at every time step, the output sequence is obtained by convolving the input sequence with the SSM kernel.

---

## A.6 Why Use the Convolution Form?

The recurrent formulation

$$
x_{k+1}=\bar{A}x_k+\bar{B}u_k
$$

requires computing

$$
x_0
\rightarrow
x_1
\rightarrow
x_2
\rightarrow
\cdots
$$

making the computation inherently sequential.

Once the kernel

$$
\bar{K}
=
(
\bar{C}\bar{B},
\bar{C}\bar{A}\bar{B},
\bar{C}\bar{A}^2\bar{B},
\ldots
)
$$

has been precomputed, the output can instead be evaluated as

$$
y=\bar{K}*u,
$$

which can be computed efficiently using highly optimized convolution algorithms (e.g., FFT-based methods) and parallel hardware such as GPUs.

This equivalence between the recurrent state-space formulation and convolution is one of the central ideas behind Structured State Space Models (S4). During training, the convolution formulation enables efficient parallel computation over the entire sequence, while preserving exactly the same linear dynamics described by the original state-space recurrence.
